```python
class Car:
    def __init__(self, brand="Unknown", model="Unknown", price=0):
        self.brand = brand
        self.model = model
        self.price = price

    def __str__(self):
        return f"{self.brand} {self.model} - ₹{self.price}"

inventory = []
def add():
  brand = input('enter brand ')
  model = input('enter model ')
  price = input('enter price ')
  cars = Car(brand,model,price)
  inventory.append(cars)
  print('car added successfully ')

def show_all_cars():
  if inventory:
    print('all cars from show room')
    for car in inventory:
      print(car)
  else:
    print('cars is not availabe in show room')

def filter_by_price():
  max_price = input('enter max price ')
  min_price = input('enter min price ')
  for car in inventory:
    if car.price >= min_price and car.price <= max_price:
      print('cars in inventory in your budget')
      print(car)
    else:
      print('not cars available in your budget')

def filter_by_brand():
    brand = input('enter brand name ')
    if inventory:
      for car in inventory:
        if car.brand == brand:
          print('car found')
          print(car)
        else:
          print('car not found')
    else:
       print('car not found, inventory is empty')



def Menu():
    while True:
        print('welcome to show room')
        print('for all cars press 1 ')
        print('for add cars press 2 ')
        print('for filter by price cars press 3 ')
        print('for filter by brand cars press 4 ')
        print('for exit press 5 ')
        key = input('choose number ')

        if key == '1':
          show_all_cars()
        elif key =='2':
          add()
        elif key == '3':
          filter_by_price()
        elif key == '4':
          filter_by_brand()
        elif key == '5':
          print('goodbye')
          break
        else:
          print('valid key 1 to 5')

Menu()
```

```python
import tkinter as tk

class Product:
  def __init__(self,product_name='item',product_price=0):
    self.price = product_price
    self.name = product_name
  def __str__(self):
    return f"{self.name} - ₹{self.price}"

class Cart:
  def __init__(self,product_name,product_price):
    self.name = product_name
    self.price = product_price
  def __str__(self):
     return f"{self.name} - ₹{self.price}"


inventory = []
cart = []
def add_product():
  name = input('enter product name')
  price = input('enter product price')
  if name and price:
    product = Product(name,price)
    inventory.append(product)
    print('product successfully added.👌')

def add_to_cart():
    name = input('Enter product name to add to cart: ')
    found = False
    for item in inventory:
        if item.name == name:
          carts = Cart(item.name,item.price)
          cart.append(carts)
          print('✅ Product added to the cart successfully.')
          found = True
          break  # stop loop after match

    if not found:
        print('❌ Failed to add to cart, product not available.')



def display_cart():
    print('🛒 Welcome to the cart section!')
    print('1. Remove a product from cart')
    print('2. Go back')
    print("\n🧾 Cart Items:")
    if cart:
        for item in cart:
            print(f"- {item}")
        total = sum(float(item.price) for item in cart)
        print(f"\n💰 Total: ₹{total}")
    else:
        print("Cart is empty 😔")
    key = input('Choose an option: ')

    if key == '1':
        name = input('Enter product name to remove: ')
        found = False
        for item in cart:
            if item.name == name:
                cart.remove(item)
                print('✅ Product removed from cart.')
                found = True
                break
        if not found:
            print('❌ Product not found in cart.')

    elif key == '2':
        display_product()
        return




def display_product():
  while True:
    print('welcome to mks store')
    print('choose 1 for add products')
    print('choose 2 for product add to the cart')
    print('choose 3 for show cart')
    print('choose 4 for exit')
    key = input('choose a number')
    if key == '1':
      add_product()
    if key == '2':
      add_to_cart()
    if key == '3':
      display_cart()

    if inventory:
      print('products List😎')
      for item in inventory:
        print(f"{item}")
    else:
      print('products is not available😔')
    if key == '4':
      print('goodbye')

display_product()





```

```python
class BankAccount:
  def __init__(self,name,account_number,balance,pin):
    self.holder_name = name
    self.holder_account_number = account_number
    self.bank_balance = balance
    self.holder_pin = pin
  def __str__(self):
    return f"{self.name} - {self.holder_account_number} - balance ₹{self.bank_balance}"

customers = []
def Menu():
  while True:
    print('welcome to the mks bank')
    print('balance check: 1')
    print('deposit money: 2')
    print('withdraw money: 3')
    print('open account: 4')
    print('exit: 0')

    key = input('choose a key')
    if key == '1':
      balance_check()
    if key == '2':
      deposit()
    if key == '3':
      withdraw()
    if key == '4':
      open_account()
    if key == '0':
      print('goodbye, have nice day !')
      break


def open_account():
  print('welcome to the mks bank')
  name = input('enter your name')
  account_number = input('enter your account number')
  balance = float(input('enter your balance'))
  pin = input('set your pin')
  user = BankAccount(name,account_number,balance, pin)
  customers.append(user)
  print('account opened successfully')

def balance_check():
  print('check balance')
  print('for exit: 1')
  pin = input('enter pin')
  found = False
  for user in customers:
    if user.holder_pin == pin:
      print('balance fetched successfully')
      print(user.bank_balance)
      found = True
    else:
      print('failed to fetched your account')

  key = input('enter a key')
  if key == '1':
    Menu()



def deposit():
  print('deposit')
  print('for exit: 1')
  pin = input('enter pin')
  found = False
  for user in customers:
    if user.holder_pin == pin:
      deposit = float(input('enter money to deposit'))
      user.bank_balance += deposit
      print('money deposited successfully')
      found = True
    else:
      print('failed to fetched your account')
  key = input('enter a key')
  if key == '1':
    Menu()


def withdraw():
  print('withdraw')
  print('for exit: 1')
  pin = input('enter pin')
  found = False
  for user in customers:
    if user.holder_pin == pin:
      deposit = float(input('enter money to withdraw'))
      user.bank_balance -= deposit
      print('money withdrawed successfully')
      found = True
    else:
      print('failed to fetched your account')

  key = input('enter a key')
  if key == '1':
    Menu()

Menu()

```

```python
class books:
  def __init__(self,book_name,book_price):
    self.name = book_name
    self.price = book_price
  def __str__(self):
    return f"{self.name} - {self.price}"

store = []
def Menu():
  while True:
    print('welcome to the book store!')
    print('add book: 1')
    print('delete book: 2')
    print('update book: 3')
    print('exit: 0')
    if store:
      for book in store:
        print(book)
    else:
      print('not books ')
    key = input('enter a key')
    if key == '1':
      add()
    if key == '2':
      delete()
    if key == '3':
      update()
    if key == '0':
      print('goodbye , have nice day ! ')
    break



def add():
  name = input('enter book name')
  price = input('enter book price')
  if name and price:
    book = books(name,price)
    store.append(book)
    print('book added successfully')
  else:
    print('please enter book name and price')
  key = input('enter 1 to exit')
  if key == '1':
    Menu()

def delete():
  name = input('enter book name which want to delete')
  found = False
  if store:
    for book in store:
      if book.name == name:
        store.remove(book)
        print('book deleted successfully')
        found = True
      else:
        print('not found book')
  key = input('enter 1 to exit')
  if key == '1':
    Menu()
def update():
  find_book = input('enter book name which want to update')
  name = input('enter book name')
  price = input('enter book price')
  found = False
  if store:
    for book in store:
      if book.name == find_book:
        store.remove(book)
        update = books(name,price)
        store.append(update)
        print('book updated successfully')
        found = True
      else:
        print('not found book')
  key = input('enter 1 to exit')
  if key == '1':
    Menu()
Menu()
```

quiz game
created in 30 minuts

```python
import random

questions = [{
    'who is the prime minister of india': 'naredra modi',
    'what is the national animal of india':'tiger',
    'Which is the national river of India?': 'ganga',
    'What is the capital of India?':'delhi',
    'India is a part of which continent?':'ashia'
} ]

score = 0
used_questions = []

# Ask all questions one by one
for _ in range(len(questions[0])):
    selected_q = random.choice([q for q in questions[0].keys() if q not in used_questions])
    used_questions.append(selected_q)

    ans = input(f"{selected_q} : ")
    correct_answer = questions[0][selected_q]

    if ans.strip().lower() == correct_answer.lower():
        print("✅ Correct Answer")
        score += 1
    else:
        print(f"❌ Wrong answer! The correct answer is: {correct_answer}")

print(f"\n🎉 Your final score is {score}/{len(questions[0])}")
```

```python
class students:
  def __init__(self,student_name,student_marks):
    self.name = student_name
    self.marks = student_marks
  def __str__(self):
    return f"name: {self.name}, marks - {self.marks}"

data = []
def add_student():
  name = input('enter name')
  count = int(input("Enter the subjects count: "))
  marks = {input("Enter subject name: "): input("Enter marks: ") for _ in range(count)}

  student = students(name,marks)
  data.append(student)
  print('student added successfully')
  key = input('enter 1 to exit')
  if key == '1':
    menu()

def search_student():
  name = input('enter student name')
  found = False
  if data:
    for student in data:
      if student.name == name:
        print(f'{student}')
        found = True
  else:
        print('not found student ')


def Update_student():
  name = input('enter student name')
  count = int(input('enter subject count'))
  marks = {input('enter the subject name: ') : input('enter the marks') for _ in range(count)}
  found = False
  if data:
    for student in data:
      if student.name == name:
        data[0] = name
        data[1]['marks'] = marks
        print('student data updated')
        found = True
      else:
        print('not found student ')


def menu():
  while True:
    print('student report card system')
    print('add student : 1')
    print('search student : 2')
    print('update student : 3')
    print('exit : 0')

    if data:
      for student in data:
        print(f"name: {student.name}")
        total_marks = 0
        for subject, mark in student.marks.items():
            print(f"  {subject}: {mark}")
            total_marks += float(mark)
        print(f"Total Marks: {total_marks}")
    else:
      print('empty')
    key = input('enter a key')
    if key == '1':
      add_student()
    if key == '2':
      search_student()
    if key == '3':
      Update_student()
    if key == '0':
      print('good bye , have nice day ')
    break

menu()
```

password manager - file handling

```python
import os
import hashlib
import json

FILE_NAME = 'users.json'
class login:
  def __init__(self,username,password):
    self.username = username
    self.password = self.hash_password(password)
  def hash_password(self,password):
    return hashlib.sha256(password.encode()).hexdigest()
  def to_dict(self):
    return {'username':self.username, 'password':self.password}


def load_users():
  if os.path.exists(FILE_NAME):
    with open(FILE_NAME,'r') as file:
      # Handle empty file case
      try:
        return json.load(file)
      except json.JSONDecodeError:
        return []
  return []

def save_users(users):
  with open(FILE_NAME, 'w') as file:
    json.dump(users, file, indent=2)



def create_account():
    print('Account creation')
    username = input('Set your username: ')
    password = input('Set your password: ')
    if password:
        confirm_password = input('Confirm your password: ')
        if password == confirm_password:
            users = load_users()
            # Check if username already exists
            if any(user['username'] == username for user in users):
              print('Account already exists, please login')
              login_user()
            else:
                user = login(username,password)
                users.append(user.to_dict())
                save_users(users)
                print('Account created successfully!')
                login_user()
        else:
            print('Passwords do not match.')
    else:
        print('Password cannot be empty.')


def login_user():
  print('login account')
  print('account creation: 1')
  username = input('enter username: ')
  password = input('enter password: ')
  users = load_users()
  entered_password_hashed = hashlib.sha256(password.encode()).hexdigest()
  found = False
  for user in users: # Iterate through the list of user dictionaries
    if user['username'] == username and user['password'] == entered_password_hashed: # Compare hashed passwords
      print(f"welcome back to dashboard {user['username']}")
      found = True
      break

  if not found:
      print('❌ please enter correct username and password')

  key = input('enter 1 to account creation or any other key to try login again: ')
  if key == '1':
    create_account()
  else:
      login_user()


login_user()
```
