```python
from collections import deque

# Create a queue
queue = deque()

# Add elements
queue.append('a')
queue.append('b')
queue.append('c')

print("Initial Queue:", queue)

# Remove elements (FIFO)
print("Removed:", queue.popleft())
print("Queue after removing:", queue)
```

```python
from queue import Queue

# Create a queue with max size
q = Queue(maxsize=4)

# Add items
q.put(1)
q.put(2)
q.put(3)
print('number',q.put(4))
print("Full:", q.full())  # True

# Remove items
print("Removed:", q.get())
print("Now size:", q.qsize())
```

```python
queue = []

# Enqueue
queue.append(10)
queue.append(20)

# Dequeue
print("Removed:", queue.pop(0))  # O(n) operation
```

```python
from queue import Queue

task = Queue(maxsize=5)
def add():
  do = input('enter task')
  task.put(do)
  print('added task')

def remove():
    if task:
      task.get()
      print('removed')

def display():
  while True:
    print('todo list')
    print('add : 1')
    print('remove : 2')
    print('exit : 0')
    if task:
      for tasks in task.queue:
        print(tasks)
    else:
      print('no task')
    key = input('enter key')
    if key == '1':
      add()
    if key == '2':
      remove()
    if key == '0':
      print('boodbye , have nice day')
      break


display()
```

```python
from queue import Queue
import uuid

q = Queue(maxsize=10)

class process:
  def __init__(self,order_id,order):
    self.id = order_id
    self.order = order

  def __str__(self):
    return f"{self.id} - {self.order}"


def order_():
  order = input('input order')
  id = uuid.uuid4()
  dict = process(id,order)
  q.put(dict)

def cancel_():
  q.get()
  print('last order canceled')


def Menu():
  while True:
    print('place order: 1')
    print('cancel order: 2')
    print('exit : 0 ')
    if q:
      for order in q.queue:
        print(order)
    else:
      print('no order')
    key = input('enter a key')
    if key == '1':
      order_()
    if key == '2':
      cancel_()
    if key == '0':
      print('goodbye')
      break

Menu()
```

#stack

```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        return "Stack is empty"

    def peek(self):
        if not self.is_empty():
            return self.items[-1]
        return "Stack is empty"

    def is_empty(self):
        return len(self.items) == 0

# Usage
s = Stack()
s.push(5)
s.push(10)
print("Top item:", s.peek())
print("Removed:", s.pop())

```

##Generator function

```python
def even_numbers(limit):
    num = 0
    while num <= limit:
        if num % 2 == 0:
            yield num
        num += 1

for val in even_numbers(10):
    print(val)
```

```python
numbers = [1,2,3,4,5]
square = list(map(lambda x: x*x, numbers))
print(square)
```

```python
numbers = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens)  # [2, 4, 6]
```

```python
people = [('Alice', 25), ('Bob', 20), ('Eve', 30)]
# Sort by age
sorted_people = sorted(people, key=lambda x: x[1])
print(sorted_people)
# [('Bob', 20), ('Alice', 25), ('Eve', 30)]
```

```python
from functools import reduce

numbers = [1, 2, 3, 4]
result = reduce(lambda x, y: x + y, numbers)
print(result)  # 10
```
